```python
import numpy as np
```

The BASICS

```python
a=np.array([1, 2, 3])
print(a)
```

```python
b=np.array([[9.0,8.0,7.0],[6.0,5.0,4.0]])
print(b)
```

```python
#To get dimension of the array we use 
print(a.ndim)
print(b.ndim)
```

```python
#To get the size of the array we use
print(a.shape)
print(b.shape)  
```

```python
# TO get  the type of the array we use
print(a.dtype)
print(b.dtype)
```

Here the size occupied by three integers are 64 so we can change the byte size as how small we need

```python
a=np.array([1, 2, 3],dtype="int16")
print(a)
```

```python
# To get  the type of the array we use
print(a.dtype)
print(b.dtype)
```

```python
#To get the item size of the array we use
print(a.itemsize)
print(b.itemsize)
```

```python
#To get the total size of the array we use
print(a.nbytes) 
```
